```python
from scipy.io import loadmat
from scipy.signal import welch
import numpy as np

fs = 128

bands = {
    "delta": (0.5, 4),
    "theta": (4, 8),
    "alpha": (8, 13),
    "beta":  (13, 30),
    "gamma": (30, 45)
}

def extract_features(eeg):

    features = []

    # Time-domain features
    for ch in eeg:
        features.extend([
            np.mean(ch),
            np.std(ch),
            np.var(ch),
            np.sqrt(np.mean(ch**2))
        ])

    # Frequency-domain features
    for ch in eeg:

        freqs, psd = welch(ch, fs=fs)

        for low, high in bands.values():

            idx = (freqs >= low) & (freqs <= high)

            band_power = np.trapezoid(
                psd[idx],
                freqs[idx]
            )

            features.append(band_power)

    return features
```

```python
sample_file = files[0]

eeg = loadmat(sample_file)["Clean_data"]

sample_features = extract_features(eeg)

print(len(sample_features))
```

```python
all_features = []

for _, row in dataset_df.iterrows():

    file_path = (
        Path("../../datasets/SAM40/filtered_data")
        / row["file"]
    )

    eeg = loadmat(file_path)["Clean_data"]

    features = extract_features(eeg)

    all_features.append(features)

print("Total recordings:", len(all_features))
print("Features per recording:", len(all_features[0]))
```

```python
features_df = pd.DataFrame(all_features)

print(features_df.shape)
features_df.head()
```

```python
final_df = pd.concat(
    [
        features_df.reset_index(drop=True),
        dataset_df[["stress_score"]].reset_index(drop=True)
    ],
    axis=1
)

print(final_df.shape)

final_df.head()
```

```python
def stress_class(score):
    if score <= 3:
        return "Low"
    elif score <= 6:
        return "Medium"
    else:
        return "High"

final_df["stress_class"] = final_df["stress_score"].apply(stress_class)

print(final_df["stress_class"].value_counts())
```

```python
final_df.to_csv(
    "../../datasets/SAM40/sam40_features.csv",
    index=False
)

print("Saved successfully")
```
