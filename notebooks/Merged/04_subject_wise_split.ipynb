```python
import numpy as np

X = np.load("../../datasets/Preprocessed/X.npy")
y = np.load("../../datasets/Preprocessed/y.npy")
subjects = np.load("../../datasets/Preprocessed/subjects.npy", allow_pickle=True)
datasets = np.load("../../datasets/Preprocessed/datasets.npy", allow_pickle=True)

print("X:", X.shape)
print("y:", y.shape)
print("subjects:", subjects.shape)
print("datasets:", datasets.shape)
```

```python
unique_subjects = np.unique(subjects)

print("Number of unique subjects:", len(unique_subjects))
print(unique_subjects)
```

```python
from sklearn.model_selection import train_test_split

# Reproducibility
RANDOM_STATE = 42

# Step 1: Train (70%) and Temp (30%)
train_subjects, temp_subjects = train_test_split(
    unique_subjects,
    test_size=0.30,
    random_state=RANDOM_STATE,
    shuffle=True
)

# Step 2: Validation (15%) and Test (15%)
val_subjects, test_subjects = train_test_split(
    temp_subjects,
    test_size=0.50,
    random_state=RANDOM_STATE,
    shuffle=True
)

print("Train Subjects:", len(train_subjects))
print("Validation Subjects:", len(val_subjects))
print("Test Subjects:", len(test_subjects))
```

```python
# Create boolean masks
train_mask = np.isin(subjects, train_subjects)
val_mask = np.isin(subjects, val_subjects)
test_mask = np.isin(subjects, test_subjects)

# Split the data
X_train, y_train = X[train_mask], y[train_mask]
X_val, y_val = X[val_mask], y[val_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("Training set:", X_train.shape, y_train.shape)
print("Validation set:", X_val.shape, y_val.shape)
print("Test set:", X_test.shape, y_test.shape)
```

```python
# Check for subject overlap

train_unique = np.unique(subjects[train_mask])
val_unique = np.unique(subjects[val_mask])
test_unique = np.unique(subjects[test_mask])

print("Train ∩ Validation:", len(set(train_unique) & set(val_unique)))
print("Train ∩ Test:", len(set(train_unique) & set(test_unique)))
print("Validation ∩ Test:", len(set(val_unique) & set(test_unique)))
```

```python
import os
import numpy as np

os.makedirs("../../datasets/Split", exist_ok=True)

np.save("../../datasets/Split/X_train.npy", X_train)
np.save("../../datasets/Split/y_train.npy", y_train)

np.save("../../datasets/Split/X_val.npy", X_val)
np.save("../../datasets/Split/y_val.npy", y_val)

np.save("../../datasets/Split/X_test.npy", X_test)
np.save("../../datasets/Split/y_test.npy", y_test)

# Save subject IDs as well
np.save("../../datasets/Split/train_subjects.npy", train_subjects)
np.save("../../datasets/Split/val_subjects.npy", val_subjects)
np.save("../../datasets/Split/test_subjects.npy", test_subjects)

print("Subject-wise split saved successfully!")
```
