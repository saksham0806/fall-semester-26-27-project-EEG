```python
import pandas as pd

dep_labels, dep_counts = np.unique(y_depression, return_counts=True)

label_map = {
    -1: "Not Applicable",
     0: "Healthy",
     1: "Depressed"
}

dep_df = pd.DataFrame({
    "Label": [label_map[i] for i in dep_labels],
    "Count": dep_counts
})

dep_df
```
