# Exploratory Data Analysis: MODMA Dataset

This notebook explores the MODMA EEG dataset for depression analysis.

```python
import os
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.append(os.path.abspath(os.path.join('..', 'src')))

from data_loader import MODMADataLoader

%matplotlib inline
```

```python
# Configuration
DATA_DIR = "/run/media/Saksham/Saksham/Code/Projects/stress_analysis_research/datasets/MODMA/EEG_3channels_resting_lanzhou_2015"
EXCEL_FILE = os.path.join(DATA_DIR, "subjects_information_EEG_3channels_resting_lanzhou_2015.xlsx")

loader = MODMADataLoader(DATA_DIR, EXCEL_FILE)
```

```python
# Load full dataset
X, y, groups = loader.load_dataset(segment_length=250*4, overlap=0)
print(f"X shape: {X.shape}, y shape: {y.shape}")
```

```python
# Visualize Class Balance
sns.countplot(x=y)
plt.title("Class Distribution (0=HC, 1=MDD)")
plt.show()
```

```python
# Visualize Raw Signals
def plot_signal(X, y, idx):
    plt.figure(figsize=(15, 3))
    for i in range(3):
        plt.plot(X[idx, :, i], label=f'Channel {i+1}')
    plt.title(f"Subject Label: {'MDD' if y[idx]==1 else 'HC'}")
    plt.legend()
    plt.show()

# Plot a few samples from each class
mdd_indices = np.where(y == 1)[0]
hc_indices = np.where(y == 0)[0]

print("MDD Sample:")
plot_signal(X, y, mdd_indices[0])

print("HC Sample:")
plot_signal(X, y, hc_indices[0])
```
